```python
import requests
from bs4 import BeautifulSoup
```

```python
# fetch and store data

page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    soup = BeautifulSoup(res.text,"lxml")    
    quotes = soup.select("div.quote")

    if not quotes:
        print("No valid page anymore")
        break;


    with open(f"scrapped_data/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page{page_count}")
        page_count+= 1
    
```

```python
# extract data 

with open("scrapped_data/quotes1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
all_quotes = soup.select("div.quote")
print(all_quotes)
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags= []
    
    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text())
    
    if "life" in all_tags:
        text = q.select_one("span.text").get_text()
        
        author = q.select_one("small.author").get_text()
        

        life_quotes.append([text,author])    
print(life_quotes)
```

```python
import pandas as pd

df = pd.DataFrame(life_quotes, columns = ["Quote", "Author"])
df
```

```python
df.to_csv("cleaned_data/quotes.csv", index = False)
```
